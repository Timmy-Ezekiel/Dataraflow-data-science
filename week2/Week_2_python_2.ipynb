````python
#Creating the BankAccount base class
class BankAccount:
    def __init__(this, owner, balance):
        this.owner = owner
        this.balance = balance
    
    def deposit(this, amount):
        if amount > 0: #Prevents the user from inputing values below zero 
            this.balance += amount
            print(f'Congratulations you just deposited ${amount}')

    def withdraw(this, amount):
        if amount > this.balance:
            print('Insufficient Balance')
        else:
            this.balance -= amount
            print(f'You just withdrew ${amount}')

#Creating the subclass SavingAccount
class SavingsAccount(BankAccount):
    def __init__(this, owner, balance, interest_rate):
        BankAccount.__init__(this, owner, balance)
        this.interest_rate = interest_rate

    def add_interest(this):
        interest = (this.balance * this.interest_rate)/100 #I used the simple interest formula (Time factor = 1 year)
        this.balance += interest

#creating an object from the class
my_account = SavingsAccount('Oluwatimilehin', 10000, 10) #The interest rate here is 10% per annum
print(f'Welcome {my_account.owner} \nAccount balance: ${my_account.balance} \nInterest Rate: {my_account.interest_rate}% per annum')
````

There are two methods that you have to implement when you create an iterator, which two?

### ANSWER: The two methods are \_\_iter\_\_ and \_\_next\_\_

````python
#Example is presented below: A geometric sequence with a common ratio of 2
class itr_clss:
    def __iter__(self):
        self.a = 5
        return self #returns the object itself
    
    def __next__(self):
        x = self.a
        self.a *= 2
        return x #Returns the next item in the sequence

itr_obj = itr_clss()

#Creating an iterator from the object
a = iter(itr_obj) 

for x in a:
    if (x > 100): #This ensures that the iteration stops after a number of time
        break
    else:
        print(x)
    
````

Which statement can be used to stop the iteration?

The statement used to stop interation is the 'StopInteration' statement, example is presented below:

````python
#To prove this, i made a child class from the itr_clss above
class stp_iter(itr_clss): 
    def __next__(self):
        if self.a < 1000:
            x = self.a
            self.a *= 2
            return x
        else:
            raise StopIteration

stp_itr = stp_iter()
b = iter(stp_itr)
for x in b:
    print(x)
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
#Create CountDown class
class CountDown:
    def __init__(self, n):
        self.n = n

    def __iter__(self):
        return self

    def __next__(self):
        if self.n == 0:
            raise StopIteration
        else:
            self.x = self.n
            self.n -= 1
            return self.x

countdwn = CountDown(10)
count_itr = iter(countdwn)

#Using for loop to print the numbers in reverse order
for x in count_itr:
    print(x)        
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
#Created my own .txt file
file_name = open('file.txt', 'w')

#Passing values of between 0 to 100 inside the file using a loop
for x in range(0, 100, 10):
    file_name.write(f'{str(x)}\n')

#Closing the file
file_name.close()

#Creating the FileLineReader class
class FileLineReader:
    def __init__(self, file_name):
        self.file_name = file_name

    def __iter__(self):
        self.x = open(self.file_name, 'r')
        return self

    def __next__(self):
        #Learnt the readline method: Returns an entire line in a file as a string
        line = self.x.readline() 

        #Checking for the end of the file
        if not line:
            self.x.close()
            raise StopIteration

        return line.strip()


#Initializing the object
my_file = FileLineReader('file.txt')
file_itr = iter(my_file)

for x in file_itr:
    print(x)
````

Create an iterator class `EvenNumbers` that takes a limit and iterates over even numbers only. Use it in a `for` loop.

````python
class EvenNumbers:
    def __init__(this, limit):
        this.limit = limit

    def __iter__(this):
        this.start = 0
        this.count = 1
        return this

    def __next__(this):
        if this.start <= this.limit:
            this.x = this.start
            this.start += 2
            this.count += 1
            return this.x
        else:
            raise StopIteration

instance = EvenNumbers(16)
even_iterator = iter(instance)

for x in even_iterator:
    print(x)
````

Create two classes: `Flyer` (method `move()` returns `"Flying"`) and `Swimmer` (method `move()` returns `"Swimming"`).  
Create a class `Duck` inheriting from both.  
- Use `super()` to decide which `move()` executes.  
- Explain the **Method Resolution Order (MRO)**.  

````python
class Flyer:
    def move(self):
        return 'Flying'
    
class Swimmer:
    def move(self):
        return 'Swimming'

class Duck(Flyer, Swimmer):
    def move(self):
        return super(Flyer, self).move()

a = Duck()
a.move()
````

### Method Resolution Order (MRO)
#### MRO is a method in which python search for a method or attribute of a class hierarchy. In cases of mutiple inheritance in python, it uses an algorithm that reads the code from left to right. This means, methods with the same name from the ```flyer``` class are picked first before the ```Swimmer```.

#### When super() is called in 'Duck' class, it looks for the next class in the MRO chain, in this case which is Flyer. So a.move() prints flying, but passing ```flyer``` and ```self``` inside the ```super() ```function, automatically move to the next class after Flyer on the MRO chain which is Swimmer.

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
    def __init__(self, n):
        self.n = n

    def __iter__(self):
        self.a, self.b = 0, 1
        return self

    def __next__(self):
        if (self.n != 0):
            current = self.a
            self.a, self.b = self.b, self.a + self.b
            self.n -=1
            return current
        else:
            raise StopIteration

a = iter(Fibonacci(10))
for x in a:
    print(x)
````

- Create subclasses:  
  - `EBook` with `file_size` and `format`.  
  - `PrintedBook` with `weight` and `dimensions`.  
- Create a base class `Member` with `name` and `member_id`.  
- Create subclasses:  
  - `StudentMember` with `grade_level`.  
  - `TeacherMember` with `department`.  
  
### 2. Polymorphism  
- Define a method `get_description()` in `Book` and override it in both `EBook` and `PrintedBook`.  
- Define a method `borrow_book()` in `Member` that behaves differently for `StudentMember` and `TeacherMember` (e.g., different borrowing limits).  
  
### 3. Iterators  
- Implement an iterator for the `Library` class to iterate over all books.  
- Implement an iterator for the `Member` class to iterate over borrowed books.  
  
### 4. Scope  
- Use private and protected attributes to demonstrate scope:  
  - `_borrowed_books` in `Member` (protected).  
  - `__library_name` in `Library` (private).  
- Provide getter and setter methods where appropriate to access private/protected attributes.  
  
### 5. Extra Features  
- Track which books are available and which are borrowed.  
- Allow searching books by title or author.  
- Count the total number of books borrowed by all members (use a class-level variable).  

````python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year

        #Once a book is created, its availability is true
        self._is_available = True

    #POLYMORPHISM: Defining a method get_description in Book class
    def get_description(self):
         return f'BOOK DETAILS\nTitle: {self.title}\nAuthor: {self.author}\nYear: {self.year}'

    
    def is_available(self):
        return self._is_available

    #This updates the availability of the book
    def set_availability(self, status):
        self._is_available = status
    
#Creating Subclasses Ebook and PrintedBook
class Ebook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    #POLYMORPHISM: Overiding get_description()
    def get_description(self):
        #I called the get_description of the base class and add it a new f'string
        x = super().get_description()
        return f'E{x}\nFile_Size:{self.file_size}MB\nFormat:{self.format}'

          
class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    #POLYMORPHISM: Overiding get_description()
    def get_description(self):
        x = super().get_description()
        return f'PRINTED {x}\nFile_Weight: {self.weight}\nDimension: {self.dimensions}'
````

````python
#Books object just to test this section of the code
book_1 = Book('Heaven Blessings', 'Ezekiel', 1963)
book_2 = Ebook('Gifted Hands', 'Ben Carson', 2010, 5.3, "PDF")
book_3 = PrintedBook('Think & Grow Rich', 'Napoleon Hill', 1963, 1.4, "25 x 18 cm")
book_4 = PrintedBook('Atomic Habit', 'James Clear', 2014, 1.4, "15 x 18 cm")

for x in [book_1, book_2, book_3]:
    print(x.get_description())
    print(f'Availability:{x._is_available}',  '\n')
````

````python
#Creating Member base class and other subclass (StudentMember and TeacherMember)
class Member:
     def __init__(self, name, member_id):
          self.name = name
          self.member_id = member_id
          #Initializing a list where i store borrow books for the user
          self._borrowed_books = []
          self.limit = 0

     total_borrowed_books = 0
     
     #POLYMORPHISM: Defining a borrow_book in member class
     def borrow_book(self, book):
          if len(self._borrowed_books) >= self.limit:
               print('Borrowing limit has been reached')
               return
          if not book.is_available():
               print('Book is not available')
               return
          
          book.set_availability(False)
          self._borrowed_books.append(book)
          Member.total_borrowed_books += 1
          return f'Hi, {self.name }. You just borrowed {book.title}'
     
     #Creating the object iterator
     def __iter__(self):
          self.index = 0
          return iter(self._borrowed_books)
     
     #Getter method
     def get_borrowed_books(self):
          return self._borrowed_books.copy()

     #Method to Return book and update availability
     def return_book(self, book):
          if book in self._borrowed_books:
               self._borrowed_books.remove(book)
               book.set_availability(True)
               Member.total_borrowed_books -= 1
               print(f'{self.name} return {book.title}')

     def get_member_details(self):
          return f'Name: {self.name} \n ID: {self.member_id}'
````

````python
#Child clases of the member class 
class StudentMember(Member):
     def __init__(self, name, member_id, grade_level):
          super().__init__(name, member_id)
          self.grade_level = grade_level

          #Updated the limit: to limit repetition of codes
          self.limit = 3

       
class TeacherMember(Member):
     def __init__(self, name, member_id, deparment):
          super().__init__(name, member_id)
          self.deparment = deparment
     
          #Updated the limit
          self.limit = 5
````

````python
class Library:
    def __init__(self, library_name):
        self.__library_name = library_name #Double underscore shows this attribute is private
        self.books = []
        self.members = []
    
    #Getter for the __library_name attribute
    def get_library_name(self):
        return self.__library_name

    #Setter for the __library_name attribute
    def set_libary_name(self, new_name):
        self.__library_name = new_name

    #Allow books to be added
    def add_book(self, book):
        self.books.append(book)

    #allow members to register
    def add_member(self, member):
        self.members.append(member)

    #Library Iterator
    def __iter__(self):
        return iter(self.books)

    #Allows user to search by title
    def search_by_title(self, title):
        results = []
        for book in self.books:
            if title.lower() == book.title.lower():
                results.append(book)
        return results
    
    def get_available_books(self):
        available_books = []

        for book in self.books:
            if book.is_available():
                available_books.append(book)
        return available_books

    def get_borrowed_books(self):
        borrowed_bks = []
        for book in self.books:
            if not book.is_available():
                borrowed_bks.append(book)
        return borrowed_bks
````
