````python
#Creating a class  that takes the list of student record and store in a json file, then return
#Json module as been imported
class json_handing:
    def __init__(self, list_obj, json_file):
        self.list_obj = list_obj
        self.json_file = json_file

    #Create a json file if not existed and stores the content of the list into it
    def store_to_json(self):
        with open(self.json_file, 'w') as file:
            file.write(json.dumps(self.list_obj))

    #Return the content of the json file to a list object
    def json_to_list(self):
        try:
            with open(self.json_file, 'r') as file:
                content = json.loads(file.read())
            return content
        except FileNotFoundError:
            print('File does not exist')

json_practice = json_handing(student_record, 'new2.json')

#Creates a json file and write the student list inside the json file
json_practice.store_to_json()

#Read the json file and return a list objec
student_info = json_practice.json_to_list()
for x in student_info:
    print(x)
````

````python
#====ASSIGNMENT 2: Math Quiz
import random

score = 0
operations = ['+', '-', '*', '/']

for i in range(5):
    num1 = random.randint(1, 20)
    num2 = random.randint(1, 20)
    op = random.choice(operations)

    if op == '+':
        answer = num1 + num2
    elif op == '-':
        answer = num1 - num2
    elif op == '*':
        answer = num1 * num2
    else:
        num1 = num1 * num2
        answer = num1 / num2

    try:
        user_answer = float(input(f"Question {i+1}: {num1} {op} {num2} = "))
    except ValueError:
        print("Invalid input, counted as incorrect.")
        continue

    if user_answer == answer:
        print("Correct!")
        score += 1
    else:
        print(f"Incorrect. The answer was {answer}.")

print(f"\nYour final score: {score}/5")
````

````python
#ASSIGNMENT 3:Error Handling System
def div():
    try:
        user_input = int(input('Enter a number: '))
        x = 100/user_input
        print(x)
    except ValueError:
        print('Invalid Input')
    except ZeroDivisionError:
        print("You can't divide by zero")
    finally:
        print('DONE')

div()

````

````python
#ASSIGNMENT 4:Date Calculator
from datetime import date

birthday_str = input("Enter your birthday (MM-DD-YYYY): ")

try:
    #Splits the user input and save to catch error in input 
    month, day, year = map(int, birthday_str.split('-'))
    today = date.today()

    # Next occurrence of the birthday
    next_birthday = date(today.year, month, day)

    # If it's already passed this year, use next year instead
    if next_birthday < today:
        next_birthday = date(today.year + 1, month, day)

    days_left = (next_birthday - today).days

    if days_left == 0:
        print("Happy birthday!")
    else:
        print(f"There are {days_left} days left until your next birthday.")

except ValueError:
    print("Please enter a valid date in MM-DD-YYYY format.")
````

### ASSESSMENT 10
#### A virtual environment is an isolated Python setup with its own installed packages, separate from your system-wide Python. It's useful because different projects often need different (sometimes conflicting) package versions, and a virtual environment keeps those dependencies contained so they don't interfere with each other or with your system Python. A new virtual environment can be created with ```python -m venv myenv``` and activated before installing packages.

````python
#ASSESSMENT 11
from datetime import date

try:
    d1 = input("Enter first date (MM-DD-YYYY): ")
    d2 = input("Enter second date (MM-DD-YYYY): ")

    m1, day1, y1 = map(int, d1.split('-'))
    m2, day2, y2 = map(int, d2.split('-'))

    date1 = date(y1, m1, day1)
    date2 = date(y2, m2, day2)

    difference = abs((date2 - date1).days)
    print(f"There are {difference} days between the two dates.")

except ValueError:
    print("Please enter valid dates in MM-DD-YYYY format.")
````
