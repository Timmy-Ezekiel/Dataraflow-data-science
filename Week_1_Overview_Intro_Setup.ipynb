````python
#TASK 4:Create a program that prints the multiplication table of a number (from 1–12).
def mul_table(num):
    for i in range(1, 13):
        print(f"{num} x {i} = {num * i}")
mul_table(5)
````

````python
#TASK 5:Write a program that calculates the factorial of a number using a loop.
def factorial(num):
    result = 1
    while (num > 0):
        result *= num
        num -= 1
    return result
factorial(5)
````

````python
#Task 5 (alternative method i also tried)
def factorial2(n):
    result = 1
    for i in range(1, n+1): #This provides a range of number from 1 to the user input
        result *= i
    return result
factorial2(5)
````

````python
 #Task 6: Write a program that finds the largest number in a given list.
def larg_num(val):
    holder = 0
    for item in val:
        if item > holder:
            holder = item
    return holder

gvn_list = [1, 2, 3, 6, 7, 8, 9, 10, 11, 12, 500, 13, 14, 15]
larg_num(gvn_list)

````

````python
#Task 7:Create a Python program that counts how many vowels are in a given string.
def check_vowel(str_inp):
    num_vowels = 0
    for char in str_inp:
        if char.lower() in ['a', 'e', 'i', 'o', 'u']:
            num_vowels += 1
    return num_vowels
check_vowel('Dataraflow')
````

````python
#TASK 8: Write a function that takes two numbers and returns their sum, difference, and product.
def run_cal(x, y):
    result = [x + y, x - y, x * y]
    return result

run_cal(15,3)
````

````python
#TASK 9:Create a program that reverses a string without using slicing ([::-1]).
my_str = 'Oluwatimilehin Ezekiel'
def reverse_str(val):
    str_len = len(val)
    for _ in range(str_len-1, -1, -1):
        print(f'{my_str[_]}', end = '')

#Calling the funtion on a string: In this case my_str
reverse_str(my_str)
````

````python
#Task 10:Finobacci Sequence: Write a program that generates the first 10 numbers of the Fibonacci sequence.
start = 0
next = 1
for _ in range(10):
    print(start, end = ' ')
    start, next = next, start + next #Tuple unpacking
````

````python
#ASSIGNMENT 1: STUDENT GRADES SYSTEM

print('Welcome!!! This is a Grading System: \nThis system takes only 5 subjects')
grades = []
#Collection of user's input and storing it in a list (grades)
for x in range(5):
    grd = float(input(f'Enter for Subject {x + 1}: '))
    #Handling cases where user inputs value below 0 or above 100
    while (grd < 0 or grd > 100):
        grd = float(input(f'Please input values between 0 and 100 for subject {x + 1}: '))
    grades.append(grd)

#Calculating the average and student grade
avg_score = sum(grades)/len(grades)
print(f'Your grades are {grades}  \nAverage score = {avg_score}')

#Placing student in a grade class
if 80 <= avg_score <= 100:
    print('A')
elif 60 <= avg_score < 80:
    print('B')
elif 40 <=  avg_score < 60:
    print('C')
else:
    print('D')
````

````python
#ASSIGNMENT 2: CALCULATOR

#----Buiding a function that takes user input and the arithmetic the user want to perform.
def calculator(*input, arr): #!!!!Learnt how to specify multiple arguments in a function. *input creates a tuple
    #Addition
    if arr == '+':
        return sum(input)
    #Subtraction
    elif arr == '-':
        result = input[0]
        for i in range(1, len(input)):
            result -= input[i]
        return result
    #Multiplication
    elif arr == '*':
        result = 1
        for i in input:
            result *= i
        return result
    #Division
    elif arr == '/':
        result = input[0]
        for i in range(1, len(input)):
            if input[i] == 0:
                return "Error: Division by zero"
            result /= input[i]
        return result
    else:
        return "Invalid operation"
calculator(45, 2, arr = '*')
````

````python
#ASSIGNMENT 3: Shopping Cart
cart = {}

while True:
    item_name = input('Enter the name of the item (or type "done" if you are finished): ')
    if item_name.lower() == 'done':
        break
    item_price = float(input(f'Enter the price of {item_name}: '))
    cart[item_name] = item_price

print('Items')
for item, price in cart.items():
    print(f'{item}: #{price:.2f}')

item_sum = sum(cart.values()) # I learnt that the values() method gives me a list of all the values in a dictionary
print(f'The total items in your cart is :{len(cart)}\nTotal price: #{item_sum}')
````

````python
#ASSIGNMENT 4: GUESSING GAME
#--- Created a random number between 1-20 importing the random module
import random
ran_num = random.randint(1, 20)
#----Taking users input and comparing it to the random number ensuring user get to guess until they get it correct
user_input = 0
while(user_input != ran_num):
    user_input = int(input('Guess the random number: '))
    if user_input == ran_num:
        print(f'Congratulations! you got it right: The random number is {ran_num}')
        break
    elif user_input > ran_num:
        print("Too High")
    else:
        print('Too Low')
````

````python
#ASSIGNMENT 5: FILE HANDLING (I/O)
#.............File opening
file = open('test.txt', 'w')

#..........Accepting user input and writing it to the file
while True:
    user_input = input('Enter text to write to file (type "exit" to quit): ')
    if user_input.lower() == 'exit':
        break
    file.write(user_input + ' ')
file.close()
````

**ASSESSMENT 8**

== compares the actual values in the comparison statment (are the contents equal?)

'is' mainly compares identity; it focuses on the memory location rather than the value

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(a == b)   # True  (same values)
print(a is b)   # False (different objects in memory)

c = a
print(a is c)   # True (c points to the same object as a)
````

````python
#ASSESSMENT 11
def second_largest(numbers):
    unique_nums = list(set(numbers)) #This handles repetition of elements by converting the list to a set, then back to a list
    unique_nums.sort()
    return unique_nums[-2]

nums = [10, 20, 4, 45, 99, 99]
print(second_largest(nums))
````

**ASSESSMENT 12**

for loop: used when you know how many times to iterate (e.g., over a sequence).

```
# This loop prints 0 to 4
for i in range(5):
    print(i)
```

while loop: it is used when looping depends on a condition being true, and the number of iterations isn't known in advance.

```
# While loop: Also prints the same values as the for loop above
i = 0
while i < 5:
    print(i)
    i += 1
```



````python
#ASSESSMENT 13
#Exceptions are errors detected during execution that can be caught and handled so the program doesn't crash.
try:
    num1 = 2
    num2 = 0
    result = num1 / num2
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
else:
    print("Result:", result)
finally:
    print("Execution complete.")
````

````python
#ASSESSMENT 14
def is_palindrome(s):
    s = s.lower().replace(" ", "")
    return s == s[::-1]

word = input("Enter a string: ")
if is_palindrome(word):
    print("It's a palindrome!")
else:
    print("Not a palindrome.")
````
